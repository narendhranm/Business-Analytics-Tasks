```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    "Price": [20, 25, 30, 35, 40],
    "Bottles Sold": [120, 110, 90, 80, 70]
}
```

```python
df = pd.DataFrame(data)
```

```python
print("Original Dataset:")
print(df)
```

```python
df["Revenue"] = df["Price"] * df["Bottles Sold"]
```

```python
print("\nUpdated Dataset with Revenue:")
print(df)
```

```python
highest_revenue = df["Revenue"].max()
```

```python
best_price = df[df["Revenue"] == highest_revenue]["Price"]
```

```python
print("\nHighest Revenue: ₹", highest_revenue)

print("Price(s) that generated the highest revenue:")
for price in best_price:
    print(f"₹{price}")
```

```python
plt.figure(figsize=(8,5))
plt.bar(df["Price"], df["Revenue"])
```

```python
plt.figure(figsize=(8,5))
plt.plot(df["Price"], df["Revenue"], marker='o', linewidth=2)
plt.title("Price vs Revenue (Line Chart)")
plt.xlabel("Price (₹)")
plt.ylabel("Revenue (₹)")
plt.xticks(df["Price"])
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
